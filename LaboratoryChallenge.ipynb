Table 1. Debate Argument

```r
dabate <- data.frame(
  ArgumentID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  Position = c("For", "For", "For", "Against", "Against", "Against"),
  Topic = c("Learning", "Productivity", "AI Literacy", "Academic Integrity", "Dependency", "Assessment"),
  Claim = c( "AI can provide students with immediate learning support.", "AI can help students complete routine academic tasks more efficiently.", "Allowing AI use can help students develop AI literacy.", "Undisclosed AI-generated work may create academic integrity concerns.", "Excessive AI use may reduce students' independent problem-solving practice.", "AI use may make it more difficult to determine whether submitted work reflects student learning." )


)

dabate
```

Table.2 Evidence

```r
evidence <- data.frame(
EvidenceID = c("E001", "E002", "E003", "E004", "E005", "E006"),
ArgumentID = c("A001", "A002", "A003", "A004", "A005", "A006"),
EvidenceType = c("Research", "Research", "Policy", "Policy", "Research", "Research"),
Strength = c("Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong" )

)

evidence
```

Exercise 5 - Create the Evidence Base

2. Count Arguments

```r
table(dabate$Position)
```

```r
nrow(dabate)
```

3. Connect Table Argument and Evidence

```r
knowledge_base <- merge(
  dabate,
  evidence,
  by = "ArgumentID"

)

knowledge_base
```

4. Convert Evidence Strength Into a Score

```r
get_score <- function(Strength)(

  if(Strength == "Strong"){
    return(3)
  } else if (Strength == "Moderate"){
    return(2)
  } else if (Strength == "Weak"){
    return(1)
  } else {
    return(0)
  }

)

get_score("Strong")
```

5. Add Evidence Score

```r
knowledge_base$EvidenceScore <- sapply(
  knowledge_base$Strenght,
  get_score
)

print(knowledge_base)
```

6. Convert Evidence Score to Support

```r
classify_support <- function(score)(

  if(score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else {
    return("No Support")
  }

)

classify_support()
```

7. Add Evidence Support

```r
knowledge_base$EvidenceSupport <- sapply(
  knowledge_base$EvidenceScore,
  classify_support
)

print(knowledge_base)
```

8. Search by Topic

```r
search_topic <- function(topic) {

  result <-subset(
    knowledge_base,
    Topic == topic
  )

  if(nrow(result) == 0){
  cat("Topic not Found")
} else {
  print(result)
  }
}

search_topic("Learning")
```
