8. Creating a Data Frame

```r
students <- data.frame(
  student_ID = c("S001","S002","S003", "S004", "S005"),
  attendance = c(95, 72, 88, 79, 68),
  grade = c(90,70,85, 77, 65)
)

students
```

9. Adding a Classication Column

```r
Adding a Classfication Column
```

```r
students$performance <- ifelse(
  students$grade >= 80,
  "Good",
  ifelse(
    students$grade >= 75,
    "Passed",
    "Failed"
  )
)

students
```

10. Creating Sample Rule-Based System

```r
students$Risk_Status <- ifelse(
  students$attendance < 75 & students$grade < 75,
  "High Risk",
  "Low Risk"
)

students
```

11. Add More Rules

```r
student_risk <- function(attendance, grade) {
  if (attendance < 75 && grade < 75) {
    return("High Risk")
  } else if (attendance < 75) {
    return("Moderate Risk (Attendance)")
  } else if (grade < 75) {
    return("Moderate Risk (Grade)")
  } else {
    return("Low Risk")
  }
}

students$Risk <- mapply(student_risk, students$attendance, students$grade)

students
```

12. Count the Result

```r
table(students$Risk)
```

13. Create a Visualization

```r
risk_counts <-table(students$Risk)

barplot(
  risk_counts,
  main= "Students Risk Classification",
  xlab = "Risk Levels",
  ylab = "Number of Students"
)
```

14. Add a Recommendation

```r
students$Recommendation <- ifelse(
  students$Risk == "High Risk",
  "Immediate Academic Intervention",
  ifelse(
    students$Risk == "Moderate Risk (Attendance)" | students$Risk == "Moderate Risk (Grade)",
    "Academic Monitoring",
    "Continue Regular Monitoring"
  )
)

students
```
